```python
import pandas as pd             

## loading the ratings into a dataframe
## having index_col='people' makes the their names as the row labels instead of just basic numbers

df = pd.read_csv('Moviesnpeople.csv', index_col='People') 

print(" Original Movei Ratings")
display(df)

## now we show the average ratings for each person and for each movie 
## axis=1 calculates the average horizontally 
## and axis=0 calculates the average vertically

print("\n Average Ratings for Each User")
display(df.mean(axis=1).round(2).reset_index(name='Average Rating'))

print("\n Average Rating by Movie")
display(df.mean(axis=0).round(2).reset_index(name='Average Rating')) 

## creating a dataframe with normalized ratings also having a min-max scrollng
## the formula is (value - min) / (max - min)
##  We transpose the dataframe (df.T) to easily do the math per user, then transpose back
df_norm = (df.T - df.T.min()) / (df.T.max() - df.T.min())
df_norm = df_norm.T

print("\n Normalized Ratings")
display(df_norm.round(2))

print("\n Average Normalized Rating by User")
display(df_norm.mean(axis=1).round(2).reset_index(name='Avg Normalized Rating'))

print("\n Average Normalized Rating by Movie")
display(df_norm.mean(axis=0).round(2).reset_index(name='Avg Normalized Rating'))
```

## Conclusions - Advantages and Disadvantages of Normalization

## Advantages:
Normalizing data (scaling it from 0 to 1) is incredibly useful for adjusting to different persons baseline grading behaviors. A harsh critic might never give a movie a 5, treating a 4 as an absolute masterpiece. Meanwhile, a generous reviewer might hand out 5s to almost everything, treating a 3 as terrible. By normalizing the data on a per person basis, we even things out. A 1.0 represents that specific persons's absolute favorite movie, and a 0.0 represents their least favorite, regardless of whether their original scores were grouped between 2-4 or 4-5. This gives a much more specific figure when representing true ratings.

## Disadvantages:
The main drawback of normalization is that it destroys the absolute magnitude of the scores. If a person watches five incredible movies and rates them all 4s and 5s, the math forces their lowest rated ones to become a 0.0. Looking only at the normalized data, an algorithm would if not tweaked enough incorrectly assume the person absolutely hated that movie, when in reality, they just liked it slightly less than a so called perfect score. Normalization exaggerates tiny differences in tightly grouped data, completely removing the context of the overall positive or negative sentiment.
